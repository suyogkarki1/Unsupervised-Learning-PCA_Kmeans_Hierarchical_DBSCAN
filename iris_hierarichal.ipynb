```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns 
from sklearn import datasets
%matplotlib inline
```

```python
iris=datasets.load_iris()
```

```python
iris_data=pd.DataFrame(iris.data)
iris_data.columns=iris.feature_names
iris_data
```

#### from data we can see we have 4 features so if we need to show the clusters isnot possible from 4 dimension means 4 features thats why we need to do first features scaling to reduce dimension we known PCA 

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_scaled=scaler.fit_transform(iris_data)
```

```python
X_scaled
```

```python
X_scaled.shape
```

```python
from sklearn.decomposition import PCA
pca=PCA(n_components=2)
```

```python
pca_scaled_data=pca.fit_transform(X_scaled)
```

```python
pca_scaled_data.shape
```

now we can see visauls of 2 dimension 

```python
plt.scatter(pca_scaled_data[:,0],pca_scaled_data[:,1])
```

for hierarichal clustering we got 2 clustering 
1. Agglomeration means combine 
2. Division means divide 

```python
import scipy.cluster.hierarchy as sc

plt.figure(figsize=(20,6 ))
plt.title('Dendogram')

sc.dendrogram(sc.linkage(pca_scaled_data,method='ward'))
#sc.linkage connects the dendogram and method='ward' tells that use eucledian distance
plt.xlabel("sample index")
plt.ylabel('Eucledian Distance')
```

```python
from sklearn.cluster import AgglomerativeClustering
cluster=AgglomerativeClustering(n_clusters=2,metric='euclidean',linkage='ward')
cluster.fit(pca_scaled_data)
```

```python
cluster.labels_
```

```python
plt.scatter(pca_scaled_data[:,0],pca_scaled_data[:,1],c=cluster.labels_)
```

now again lets check metrics silouethe score

```python
from sklearn.metrics import silhouette_score

silhouette=[]
for k in range(2,11):
    algo=AgglomerativeClustering(n_clusters=k,metric='euclidean',linkage='ward')
    algo.fit(X_scaled)
    score=silhouette_score(X_scaled,algo.labels_)
    silhouette.append(score)
```

```python
silhouette
```

```python
plt.plot(range(2,11),silhouette)
plt.xticks(range(2,11))
plt.xlabel('Number of clusters')
plt.ylabel('silhouette')
plt.show()
```

```python

iris_data['cluster'] = cluster.labels_

cluster_summary = iris_data.groupby('cluster').mean()
print(cluster_summary)
```

```python
import seaborn as sns

sns.pairplot(iris_data, hue='cluster', palette='Set1')
plt.show()
```

```python
iris_data.groupby('cluster').mean()
```
